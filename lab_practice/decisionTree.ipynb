## Decision Tree: Calculate Gini Impurity

## Import library and dataset

```python
### import dataset
import pandas as pd
import numpy as np
data = {
    "Age": ["Young", "Adult", "Young", "Senior", "Young", "Adult", "Young", "Senior", "Adult", "Senior", "Adult", "Senior", "Young", "Adult", "Senior"],
    "BMI": ["Normal", "Obese", "Normal", "Obese", "Obese", "Normal", "Overweight", "Overweight", "Obese", "Overweight", "Overweight", "Normal", "Overweight", "Normal", "Overweight"],
    "History": ["None", "Severe", "Severe", "None", "None", "Severe", "Mild", "None", "None", "Mild", "None", "Severe", "Mild", "Mild", "Severe"],
    "Exercise": ["High", "Medium", "Medium", "Low", "High", "High", "Low", "High", "High", "Medium", "High", "Medium", "Medium", "Low", "Medium"],
    "Risk": ["Low", "High", "Low", "High", "Low", "High", "High", "Low", "Low", "High", "Low", "Low", "Low", "High", "High"]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Find Gini Index for the target feature
def calculate_gini_impurity(values):
    unique_values, counts = np.unique(values, return_counts=True)
    probabilities = counts / len(values)
    gini = 1 - np.sum(probabilities ** 2) 
    return gini

# Gini Impurity based on features: Age(Young, Adult, Senior)
gini_young = calculate_gini_impurity(df[df["Age"] == "Young"]["Risk"])
gini_adult = calculate_gini_impurity(df[df["Age"] == "Adult"]["Risk"])
gini_senior = calculate_gini_impurity(df[df["Age"] == "Senior"]["Risk"])
gini_age = (
    len(df[df["Age"] == "Young"]) / len(df) * gini_young +
    len(df[df["Age"] == "Adult"]) / len(df) * gini_adult +
    len(df[df["Age"] == "Senior"]) / len(df) * gini_senior
)
print("The gini weighted impurity for Age is:", round(gini_age, 4))

## Gini Impurity based on features: BMI(Normal, Overweight, Obese)
gini_normal = calculate_gini_impurity(df[df["BMI"] == "Normal"]["Risk"])
gini_overweight = calculate_gini_impurity(df[df["BMI"] == "Overweight"]["Risk"])
gini_obese = calculate_gini_impurity(df[df["BMI"] == "Obese"]["Risk"])
gini_bmi = (
    len(df[df["BMI"] == "Normal"]) / len(df) * gini_normal +
    len(df[df["BMI"] == "Overweight"]) / len(df) * gini_overweight +
    len(df[df["BMI"] == "Obese"]) / len(df) * gini_obese
)
print("The gini weighted impurity for BMI is:", round(gini_bmi, 4))

## Gini Impurity based on features: History(None, Mild, Severe)
gini_none = calculate_gini_impurity(df[df["History"] == "None"]["Risk"])
gini_mild = calculate_gini_impurity(df[df["History"] == "Mild"]["Risk"])
gini_severe = calculate_gini_impurity(df[df["History"] == "Severe"]["Risk"])
gini_history = (
    len(df[df["History"] == "None"]) / len(df) * gini_none +
    len(df[df["History"] == "Mild"]) / len(df) * gini_mild +
    len(df[df["History"] == "Severe"]) / len(df) * gini_severe
)
print("The gini weighted impurity for History is:", round(gini_history, 4))

## GIni Impurity based on features: Exercise(Low, Medium, High)
gini_exercise_low = calculate_gini_impurity(df[df["Exercise"] == "Low"]["Risk"])
gini_exercise_medium = calculate_gini_impurity(df[df["Exercise"] == "Medium"]["Risk"])
gini_exercise_high = calculate_gini_impurity(df[df["Exercise"] == "High"]["Risk"])
gini_exercise = (
    len(df[df["Exercise"] == "Low"]) / len(df) * gini_exercise_low +
    len(df[df["Exercise"] == "Medium"]) / len(df) * gini_exercise_medium +
    len(df[df["Exercise"] == "High"]) / len(df) * gini_exercise_high
)
print("The gini weighted impurity for Exercise is:", round(gini_exercise, 4))
```

```python
gini_scores = {
    "Age": gini_age,
    "BMI": gini_bmi,
    "History": gini_history,
    "Exercise": gini_exercise
}

best_feature = min(gini_scores, key=gini_scores.get)

print("Weighted Gini:")
for feature, score in gini_scores.items():
    print(f"{feature}: {score:.4f}")

print("Best feature to split:", best_feature)
```
