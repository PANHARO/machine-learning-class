<details>
<summary>Click to see more about <span style="color:orange;">df.duplicated()</span></summary>

The `df.duplicated()` function checks each row in the DataFrame and returns `True` if the row is a duplicate of a previous row, and `False` otherwise.  
When combined with `.sum()`, it counts the total number of duplicate rows in the dataset.  

This is useful to identify duplicate entries that may need to be **removed** to ensure accurate analysis.

For more details and examples, visit the official Pandas documentation:  
[Learn more about `df.duplicated()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.duplicated.html)

</details>

##### 6.2 Barplot: Compare number posts for each jobs


````python
# Number of posts for each jobs

'''
df["job_title"].value_counts()
df["job_title"].value_counts().plot(kind="bar")'''


# Count of each job title
job_counts = df["job_title"].value_counts()

# x = job titles, y = counts
x = job_counts.index # job-titles
y = job_counts.values # job-count

# Plot bar chart
plt.figure(figsize=(10, 6))
sns.barplot(x=y, y=x)  # horizontal bar chart
#sns.barplot(x=x, y=y)
plt.title("Number of Job Postings by Job Title")
plt.xlabel("Number of Jobs")
plt.ylabel("Job Title")
plt.show()
````

##### 6.3 Bar Chat: Compare which job has the highest average salary



````python

# 1. Group by job title and calculate average salary
avg_salary_by_job = df.groupby("job_title")["salary_usd"].mean()

# 2. Sort from highest to lowest
avg_salary_by_job_sorted = avg_salary_by_job.sort_values(ascending=False)

# 3. Display the top 5 highest-paying job titles
print(avg_salary_by_job_sorted.head(5))

# Plot graph
# Bar chart of top 5 highest-paying job titles
plt.figure(figsize=(10, 6))
sns.barplot(x=avg_salary_by_job_sorted.head(5).values,
            y=avg_salary_by_job_sorted.head(5).index,
            palette="viridis")
plt.title("Top 5 Highest-Paying Job Titles")
plt.xlabel("Average Salary (USD)")
plt.ylabel("Job Title")
plt.show()
````

##### 6.4 Distribution of salary currencies

````python
# Counts how many jobs are listed in each currency.

# Count number of jobs per currency
currency_counts = df["salary_currency"].value_counts()

# x = categories (currencies), y = counts
x = currency_counts.index      # ['USD', 'EUR', 'GBP', ...]
y = currency_counts.values     # [300, 120, 50, ...]

# Create bar chart
plt.figure(figsize=(6,4))
plt.bar(x, y, color="blue")   # x = categories, y = counts
plt.title("Number of Job Postings by Salary Currency")
plt.xlabel("Currency")
plt.ylabel("Number of Jobs")
plt.show()
````

### Q3. Plot a graph to compare which job titles require Python,Java, and R programming skills

````python
### Q3. Plot a graph to compare which job titles require Python,Java, and R programming skills
def Q3(df):
    # TO DO:
    python_dix = df["required_skills"].str.contains("Python")
    job_with_Python = df["job_title"][python_dix].value_counts()
    
    
    java_dix = df["required_skills"].str.contains("Java")
    job_with_Java = df["job_title"][java_dix].value_counts()
    

    r_dix = df["required_skills"].str.contains("R")
    job_with_R = df["job_title"][r_dix].value_counts()

    # Combine the results into a DataFrame for plotting
    
    combined_df = pd.DataFrame({
        'Python': job_with_Python,
        'Java': job_with_Java,
        'R': job_with_R
    }).fillna(0)  
    
    combined_df.plot(kind='bar', figsize=(12, 6))
    plt.title('Number of Job Postings Requiring Python, Java, and R Skills')
    plt.xlabel('Job Title')
    plt.ylabel('Number of Job Postings')
    plt.xticks(rotation=45, ha='right')
    plt.legend(title='Programming Language')
    plt.tight_layout()
    plt.show()

Q3(df)
````

### Q6. Education Level and Salary
The dataset contains an education_required column with four levels: Associate, Bachelor, Master, and PhD.

- Compute the `average salary` for each education level
- Visualize the result using a bar chart
- Does a higher education level always lead to a higher salary? Explain your observation.

````python
def Q6(df):
    # TO DO:
    education_salary = df.groupby("education_required")["salary_usd"].mean()
    # computation
    print("Average salary for each education level: \n", education_salary)
    # bar chart
    education_salary.plot(kind='bar', figsize=(8, 5))
    plt.title('Average Salary by Education Level')
    plt.xlabel('Education Level')
    plt.ylabel('Average Salary (USD)')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
Q6(df)
````

### Higher education level does not always lead to higher salaries. As we can see from the data set, the mean of those with associate and bachelor degree are higher than people with phd degree. This could be due to job roles, company salary variation, and experience lebels that may not always be correlated directly with education level.

### Q7. Industry and Salary
There are 15 industries in this dataset.

- Compute the `average salary` for each industry
- Sort from highest to lowest and display only the top 5
- Visualize using a horizontal bar chart

````python
def Q7(df):
    # TO DO:
    # console
    industry_salary = df.groupby("industry")["salary_usd"].mean().sort_values(ascending=False).head(5)
    print("Average salary for each industry (top 5): \n", industry_salary)
    # graph
    industry_salary[::-1].plot(kind='barh', figsize=(8, 5))
    plt.title('Average Salary by Industry (Top 5)')
    plt.xlabel('Average Salary (USD)')
    plt.ylabel('Industry')
    plt.tight_layout()
    plt.show()
Q7(df)
````

### Q9. Which job title has the highest number of fully remote (100%) positions?

### Q10. What types of AI jobs in the United States require a Bachelor’s degree and are offered by large companies?

````python
def Q10(df):
    # TO DO:
    us_dix = df["company_location"] == "United States"
    bachelor_dix = df["education_required"].str.contains("Bachelor", na=False)
    large_company_dix = df["company_size"] == "L"
    
    filtered_jobs = df[us_dix & bachelor_dix & large_company_dix]
    job_titles = filtered_jobs["job_title"].unique()
    
    print("Types of AI jobs in the United States that require a Bachelor’s degree and are offered by large companies: \n", job_titles)

Q10(df)
````
