```python
import pandas as pd
from math import prod

# Create dataset
data = {
    'Outlook': ['Rainy', 'Rainy', 'Overcast', 'Sunny', 'Sunny', 'Sunny', 'Overcast', 'Rainy', 
                'Rainy', 'Sunny', 'Rainy', 'Overcast', 'Overcast', 'Sunny'],
    'Temperature': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool', 'Cool', 'Cool', 'Mild',
                    'Cool', 'Mild', 'Mild', 'Mild', 'Hot', 'Mild'],
    'Humidity': ['High', 'High', 'High', 'High', 'Normal', 'Normal', 'Normal', 'High',
                 'Normal', 'Normal', 'Normal', 'High', 'Normal', 'High'],
    'Windy': [False, True, False, False, False, True, True, False,
              False, False, True, True, False, True],
    'Play': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'No',
             'Yes', 'Yes', 'Yes', 'Yes', 'Yes', 'No']
}

df = pd.DataFrame(data)

# Function to calculate prior probabilities
def calculate_prior(df, class_col):
    classes = df[class_col].unique()
    priors = {}
    for c in classes:
        priors[c] = len(df[df[class_col] == c]) / len(df)
    

# Function to calculate likelihood for a given feature value
def calculate_likelihood(df, feature, value, target, target_class):
    subset = df[df[target] == target_class]
    likelihood = len(subset[subset[feature] == value]) / len(subset)
    return likelihood

# Function to predict the class for a new sample
def predict(sample, df, target):
    priors = calculate_prior(df, target)
    classes = priors.keys()
    probs = {}

    for c in classes:
        prob = priors[c]
        for feature, value in sample.items():
            prob *= calculate_likelihood(df, feature, value, target)
        probs[c] = prob

    # Normalize probabilities
    total = sum(probs.values())
    for c in probs:
        probs[c] = probs[c] / total if total > 0 else 0

    return max(probs), probs

# Example test data
test_sample = {'Outlook': 'Overcast', 'Temperature': 'Mild', 'Humidity': 'Normal', 'Windy': False}

# Predict
predicted_class, probabilities = predict(test_sample, df, 'Play')
print("Predicted Class:", predicted_class)
print("Class Probabilities:", probabilities)
```
